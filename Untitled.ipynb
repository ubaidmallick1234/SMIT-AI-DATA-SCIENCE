# Question No. 07

```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operator = input("Enter operator (+, -, x, /): ")

if operator == "+":
    print("Result:", num1 + num2)
elif operator == "-":
    print("Result:", num1 - num2)
elif operator == "x":
    print("Result:", num1 * num2)
elif operator == "/":
    if num2 != 0:
        print("Result:", num1 / num2)
    else:
        print("Cannot divide by zero.")
else:
    print("Invalid operator.")
```

# Question No. 11

```python
temperature = float(input("Enter temperature in Celsius: "))

if temperature < 0:
    print("Freezing")
elif temperature > 0 and temperature < 26:
    print("Moderate")
elif temperature > 26:
    print("Hot")
else:
    print("Temperature is 0 or 26 degrees.")
```

# Question No. 12

```python
salary = float(input("Enter your salary: Rs. "))

if salary > 600000:
    tax = salary * 0.05
    gross_salary = salary - tax
    print("Tax:", tax)
    print("Gross Salary:", gross_salary)
else:
    print("Gross Salary:", salary)
```

# Question No. 13

```python
salary = float(input("Enter your salary: Rs. "))

if salary > 2500000:
    tax = salary * 0.15
elif salary >= 1500000:
    tax = salary * 0.10
elif salary >= 500000:
    tax = salary * 0.05
else:
    tax = 0

gross_salary = salary - tax

print("Tax Deduction: Rs.", tax)
print("Gross Salary: Rs.", gross_salary)
```
